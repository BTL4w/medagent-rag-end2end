# 08_THREAD_ID_CONTINUOUS_CHAT

Demo chat lien tuc voi cung `thread_id` de su dung memory checkpoint (PostgreSQL) trong LangGraph.

```python
import os
import sys

sys.path.append(os.path.abspath(".."))

from dotenv import load_dotenv
from src.agents.orchestrator import run_agent

load_dotenv()

if not os.getenv("MEMORY_URL"):
    raise ValueError("Missing MEMORY_URL in .env")
```

```python
os.getenv("MEMORY_URL")
```

```python
from langgraph.checkpoint.postgres import PostgresSaver

with PostgresSaver.from_conn_string(os.getenv("MEMORY_URL")) as cp:
    cp.setup()
```

```python
thread_id = "demo_user_001"

conversation1 = [
    "Chào bạn, tôi là Nam.",
    "Nguyên nhân của bênh sốt xuất huyết là gì?",
    "Bạn nhắc lại tên tôi là gì?",
    "Triệu chứng tôi vừa nói là gì?"
]

```

```python
thread_id = "demo_user_002"

conversation2 = [
    "Chào bạn, tôi là Nam.",
    "Bạn có thể làm được những gì?",
    "Tên của tôi là gì?"
]
```

```python
for i, user_query in enumerate(conversation2, start=1):
    print("\n" + "=" * 60)
    print(f"Turn {i}")
    print("User:", user_query)

    response = run_agent(user_query, thread_id=thread_id)
    print("Assistant:", response)
```

```python
# Thu lai voi cung thread_id de tiep tuc context cu
follow_up = "Neu toi muon dat lich kham, toi can cung cap gi?"
print(run_agent(follow_up, thread_id=thread_id))
```

```python
# Neu doi thread_id, hoi thoai se duoc tach memory
new_thread_id = "demo_user_002"
print(run_agent("Ban con nho ten toi khong?", thread_id=new_thread_id))
```
